## 3. Feature Engineering

Five features are engineered from the raw dataset to capture key customer behaviours relevant to segmentation: **Frequency**, **Recency**, **CLV** (Customer Lifetime Value), **Avg_Unit_Cost**, and **Age**.

```python
# Helper function: convert currency strings to float
def currency_to_float(val):
    val = str(val).replace('$', '').replace(',', '').strip()
    if val.startswith('(') and val.endswith(')'):
        return -float(val[1:-1])
    try:
        return float(val)
    except:
        return np.nan

# Convert currency columns
data['Total Revenue'] = data['Total Revenue'].apply(currency_to_float)
data['Unit Cost'] = data['Unit Cost'].apply(currency_to_float)

# Parse dates
data['Delivery_Date'] = pd.to_datetime(data['Delivery_Date'], format='%d%b%Y')
data['DOB'] = pd.to_datetime(data['Customer_BirthDate'], format='%d%b%Y')

# Feature: Frequency (number of orders per customer)
data['Frequency'] = data.groupby('Customer ID')['Order ID'].transform('count')

# Feature: Recency (days since last delivery)
reference_date = pd.Timestamp('2016-12-30')
data['Recency'] = (reference_date - data['Delivery_Date']).dt.days

# Feature: CLV (total revenue per customer, summed across orders)
data['CLV'] = data.groupby('Customer ID')['Total Revenue'].transform('sum')

# Feature: Avg_Unit_Cost
data['Avg_Unit_Cost'] = data.groupby('Customer ID')['Unit Cost'].transform('mean')

# Feature: Age
data['Age'] = (reference_date - data['DOB']).dt.days // 365

print("Features created successfully.")
data[['Customer ID', 'Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']].head()
```

The five features are successfully computed for each row. *Frequency* reflects the total number of orders per customer, *Recency* is the number of days since the last delivery relative to 30 December 2016 (the final date in the dataset, used as a fixed reference for reproducibility), *CLV* is the total revenue generated by each customer, *Avg_Unit_Cost* is the mean unit cost across all orders, and *Age* is derived from the customer's birth date.

### 3.1 Aggregation: One Row per Customer

```python
# Select relevant columns and aggregate to one row per customer
df_selected = data[['Customer ID', 'Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']].copy()

agg_operations = {
    'Frequency': 'first',
    'Recency': 'min',
    'CLV': 'first',
    'Avg_Unit_Cost': 'first',
    'Age': 'first'
}

df_aggr = df_selected.groupby('Customer ID').agg(agg_operations).reset_index()

print(f"Aggregated shape: {df_aggr.shape}")
df_aggr.head()
```

After aggregation, the dataset is reduced to **68,300 unique customers**, each represented by a single row with the five engineered features. Recency is aggregated using the minimum value (most recent delivery), while Frequency, CLV, Avg_Unit_Cost, and Age are carried forward from the pre-computed per-customer values.

## 5. Feature Scaling

K-means and hierarchical clustering are distance-based algorithms sensitive to feature scale. StandardScaler is applied to standardise all five features to zero mean and unit variance before clustering.

```python
features = ['Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_aggr[features])
X_scaled = pd.DataFrame(X_scaled, columns=features)

print("Scaled data — mean (should be ~0) and std (should be ~1):")
print(X_scaled.describe().round(3).loc[['mean', 'std']])
```

StandardScaler successfully normalises all five features to zero mean and unit standard deviation, ensuring that no single feature dominates the distance calculations in the clustering algorithms due to differences in scale.

## 5.1 Pipeline Implementation

Following the project guidance, a Pipeline combining StandardScaler and KMeans is implemented as a more efficient alternative. This encapsulates the preprocessing and modelling steps into a single object, reducing the risk of data leakage and improving code reproducibility.

```python
from sklearn.pipeline import Pipeline

features = ['Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('kmeans', KMeans(n_clusters=5, init='k-means++', n_init=10, random_state=42))
])

pipeline.fit(df_aggr[features])

print("Pipeline fitted successfully.")
print(pipeline)
```

## 6. Determining the Optimal Number of Clusters (k)
### 6.1 Elbow Method

```python
wcss = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(9, 5))
plt.plot(k_range, wcss, marker='o', color='steelblue', linewidth=2)
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS')
plt.title('Elbow Method — Optimal k')
plt.xticks(k_range)
plt.grid(True, alpha=0.3)
plt.show()

for k, w in zip(k_range, wcss):
    print(f"k={k}: WCSS = {w:.2f}")
```

### 6.2 Silhouette Score

```python
sil_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels, sample_size=10000, random_state=42))

plt.figure(figsize=(9, 5))
plt.plot(k_range, sil_scores, marker='o', color='darkorange', linewidth=2)
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score — Optimal k')
plt.xticks(k_range)
plt.grid(True, alpha=0.3)
plt.show()

for k, s in zip(k_range, sil_scores):
    print(f"k={k}: Silhouette = {s:.4f}")
```

### 6.3 Davies-Bouldin Index

```python
dbi_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    dbi_scores.append(davies_bouldin_score(X_scaled, labels))

plt.figure(figsize=(9, 5))
plt.plot(k_range, dbi_scores, marker='o', color='seagreen', linewidth=2)
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Davies-Bouldin Index')
plt.title('Davies-Bouldin Index — Optimal k')
plt.xticks(k_range)
plt.grid(True, alpha=0.3)
plt.show()

for k, d in zip(k_range, dbi_scores):
    print(f"k={k}: DBI = {d:.4f}")
```

The Davies-Bouldin Index (DBI) measures the average similarity between each cluster and its most similar neighbour (lower values indicate better-separated clusters). The DBI reaches its minimum at k=6 (1.1289), with k=5 scoring marginally higher at 1.1335. Given the negligible difference between these two values, and that both the elbow method and silhouette score independently favour k=5, this near-equivalence supports k=5 as the final choice. Selecting k=5 over k=6 also yields more interpretable and actionable customer segments for the business.

### 6.4 Choosing the Optimal k

Three complementary methods are used to determine the optimal number of clusters. The elbow method shows a marked deceleration in WCSS reduction after k=5. The silhouette score reaches its peak at k=5 (0.2649). The Davies-Bouldin Index is minimised at k=6 (1.1289), with k=5 scoring near-identically at 1.1335. The convergence of all three methods on k=5 and the negligible gain of k=6, which provides strong justification for selecting **k=5** as the optimal number of customer segments.

The overlap between clusters in 2D is expected given that PCA is a linear method and the clusters were defined in five-dimensional space.

### 9.2 t-SNE (t-Distributed Stochastic Neighbour Embedding)

```python
# Sample 10,000 for speed.
sample_idx = df_aggr.sample(n=10000, random_state=42).index
X_tsne_input = X_scaled.loc[sample_idx]
cluster_sample = df_aggr.loc[sample_idx, 'Cluster']

tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000)
X_tsne = tsne.fit_transform(X_tsne_input)

plt.figure(figsize=(10, 7))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1],
                      c=cluster_sample, cmap='tab10',
                      alpha=0.5, s=10)
plt.colorbar(scatter, label='Cluster')
plt.title('t-SNE — Customer Segments in 2D')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.tight_layout()
plt.show()
```

Unlike PCA, t-SNE is a non-linear dimensionality reduction method that preserves local structure, grouping similar customers together in 2D space regardless of global geometry. The t-SNE visualisation (10,000 customers sampled for computational efficiency) reveals considerably clearer cluster separation than PCA: Cluster 2 (Young Active Customers, green) and Cluster 0 (Older Loyal Customers, blue) form distinct regions on the left and bottom respectively, while Cluster 4 (Premium Infrequent Buyers, grey) is visibly separated to the right. Cluster 1 (Churned Customers) and Cluster 3 (VIP Champions) show some overlap in the centre, which is consistent with their similar frequency and unit cost profiles despite differing in recency and CLV. Overall, t-SNE provides stronger visual evidence of meaningful cluster structure than PCA, confirming that the five segments capture genuine patterns in the customer data.
